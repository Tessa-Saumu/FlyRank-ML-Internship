# Successive warehouse feature experiments

This notebook runs a fresh, explicit experiment configuration against the gated `FlyRank/internship-warehouse` release. It downloads revision-pinned Parquet shards from the Hugging Face Hub into the standard Hugging Face cache, then uses DuckDB to aggregate them locally; it does not use the local starter CSV or save raw records. June is excluded from all development reads.

The runner compares the existing five features with the momentum group (velocity, short-term acceleration, prior-coverage flag) across December 2025–April 2026 prediction origins, using mature latest-only, latest-three, and expanding training histories. It compares LR and the existing RF, random selection, the unchanged static rule, and predeclared traffic rules. It reports unknown outcome bounds, coverage and client concentration diagnostics, a zero-day sensitivity, and conditional feature-group ablations.

The default seven-day reporting delay is a declared scenario, not a verified warehouse SLA. Feature windows end seven days before each prediction date in that scenario; labels train only after their full 30-day window plus that delay has elapsed. Run this notebook top to bottom in a fresh kernel. Structured aggregate-only results are generated under `work/outputs/warehouse_feature_experiments/`; no client/page identifiers or row-level features are saved.

```python
from pathlib import Path
import runpy

_candidates = [Path.cwd(), *Path.cwd().parents]
_repo_root = next(
    (
        candidate
        for candidate in _candidates
        if (candidate / "work" / "scripts" / "warehouse_feature_experiments.py").is_file()
    ),
    None,
)
if _repo_root is None:
    raise FileNotFoundError(
        "Could not locate work/scripts/warehouse_feature_experiments.py from the notebook working directory."
    )

_runner = runpy.run_path(
    str(_repo_root / "work" / "scripts" / "warehouse_feature_experiments.py"),
    run_name="warehouse_feature_experiments",
)
_run_configuration = _runner["ExperimentConfig"]()
_runner["run_experiment"](_run_configuration)
```
